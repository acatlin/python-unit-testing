## The check that disappears

A small helper that rescales a numeric column so its smallest value becomes 0 and its largest becomes 1. Run it and look at the result.

```python
import pandas as pd

def rescale01(x):
    lo, hi = x.min(), x.max()
    return (x - lo) / (hi - lo)

rescale01(pd.Series([0, 5, 10]))
```

## Say what you expect

The same check written down as an assertion. Run it: nothing is printed. **Silence means the assertion passed.**

```python
assert rescale01(pd.Series([0, 5, 10])).tolist() == [0, 0.5, 1]
```

## Watch one fail

The expected value is wrong on purpose, so **this cell is meant to fail**. A notebook lays the error out differently from the console shown on the lesson page; what matters is the same: it ends with `AssertionError`.

If you chose Run all, Colab stops at this cell. Carry on from the next one by hand.

```python
assert rescale01(pd.Series([0, 5, 10])).tolist() == [0, 0.5, 2]
```

Python does not show the two values that disagreed. To see the actual value, run the left side by itself.

```python
rescale01(pd.Series([0, 5, 10])).tolist()
```

## Equal, or close enough?

```python
0.1 + 0.2 == 0.3
```

```python
rescale01(pd.Series([0.1, 0.2, 0.3])).tolist()
```

With `pytest.approx()` around the expected value, `==` ignores tiny floating-point differences. This cell is silent: it passes.

```python
import pytest

assert rescale01(pd.Series([0.1, 0.2, 0.3])).tolist() == pytest.approx([0, 0.5, 1])
```

The same check with a bare `==`. **This cell is meant to fail**, for a reason that has nothing to do with a bug.

```python
assert rescale01(pd.Series([0.1, 0.2, 0.3])).tolist() == [0, 0.5, 1]
```
